## Question 1 - Lists

Take your roll number, extract its individual digits, and multiply each digit by 10 to form list L.

```python
roll_no = "1024170090"

L = [int(d) * 10 for d in roll_no]
print("L =", L)
```

## Question 2 - Tuples

Create a tuple named scores containing the first 8 values from list L.

```python
L_original = [int(d) * 10 for d in "1024170090"]
scores = tuple(L_original[:8])

print("scores =", scores)
```

**(i)** Find the highest score and its index, and the lowest score and how many times it appears.

```python
highest = max(scores)
lowest = min(scores)

print("Highest score:", highest, "at index", scores.index(highest))
print("Lowest score :", lowest, "occurs", scores.count(lowest), "times")
```

**(ii)** Reverse the tuple and return the result as a list.

```python
reversed_list = list(scores[::-1])
print("Reversed:", reversed_list)
# Tuples are immutable, so they have no reverse() method - slicing builds a new object instead
```

**(iii)** Ask the user for a score and print its first occurrence index if present.

```python
x = int(input("Enter a score to search: "))

if x in scores:
    print(x, "found at index", scores.index(x))
else:
    print(x, "is not present in the tuple")
```

**(iv)** Attempt to change one element of the tuple directly and capture the error.

```python
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
# Tuples are immutable so item assignment is not allowed; a list would accept this without error
```

**(v)** Unpack the tuple into a first score, a second score, and all remaining scores.

```python
first, second, *rest = scores

print("First :", first)
print("Second:", second)
print("Rest  :", rest)
```

## Question 3 - Random Numbers

Set the random seed to your roll number before generating the numbers.

**(i)** Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
import random

random.seed(1024170090)
numbers = [random.randint(100, 900) for _ in range(100)]

print(numbers)
```

**(ii)** Count and print all odd numbers in the list.

```python
odd = [n for n in numbers if n % 2 != 0]

print("Count of odd numbers:", len(odd))
print(odd)
```

**(iii)** Count and print all even numbers in the list.

```python
even = [n for n in numbers if n % 2 == 0]

print("Count of even numbers:", len(even))
print(even)
```

**(iv)** Count and print all prime numbers in the list.

```python
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True


primes = [n for n in numbers if is_prime(n)]

print("Count of prime numbers:", len(primes))
print(primes)
```

**(v)** Print the number that occurs most frequently and how many times it occurs.

```python
most_common = max(set(numbers), key=numbers.count)

print("Most frequent number:", most_common)
print("It occurs", numbers.count(most_common), "times")
```

## Question 4 - Sets

Using the first 8 digits of the roll number, create set A = {digit x 7} and set B = {digit x 9}.

```python
digits = [int(d) for d in "1024170090"[:8]]

A = {d * 7 for d in digits}
B = {d * 9 for d in digits}

print("Digits used:", digits)
print("A =", A)
print("B =", B)
```

**(vi)** Find the union of A and B.

```python
print("Union:", A.union(B))
```

**(vii)** Find the intersection of A and B.

```python
print("Intersection:", A.intersection(B))
```

**(viii)** Find A - B and B - A separately using difference().

```python
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
# difference() gives elements of one set only, while symmetric_difference() gives both differences combined
```

**(ix)** Find the symmetric difference of A and B.

```python
print("Symmetric difference:", A.symmetric_difference(B))
```

**(x)** Check whether A is a subset of B, and whether B is a superset of A.

```python
print("A is a subset of B  :", A.issubset(B))
print("B is a superset of A:", B.issuperset(A))
```

**(xi)** Ask the user for a value X and remove it from set A using discard().

```python
X = int(input("Enter a value to remove from A: "))

A.discard(X)
print("A after discard:", A)
# discard() does nothing if the value is absent, while remove() would raise a KeyError
```

## Question 5 - Dictionaries

```python
my_dict = {
    "name": "Jasroop Singh Sachdeva",
    "roll_no": "1024170090",
    "branch": "Computer Science and Engineering",
    "age": 20,
    "city": "Ludhiana"
}

print(my_dict)
```
