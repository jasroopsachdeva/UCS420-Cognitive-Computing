### Assignment 5.3
WAP to add all prime numbers from 1 to n and n is given by the user.

```python
n = int(input("Enter n: "))

total = 0
for num in range(2, n + 1):
    is_prime = True
    for j in range(2, num):
        if num % j == 0:
            is_prime = False
            break
    if is_prime:
        total = total + num

print("Sum of primes =", total)
```

### Assignment 6.1
WAP using function that add all odd numbers from 1 to n, n is given by the user.

```python
def sum_odd(n):
    total = 0
    for i in range(1, n + 1):
        if i % 2 != 0:
            total = total + i
    return total


n = int(input("Enter n: "))
print("Sum of odd numbers =", sum_odd(n))
```

### Assignment 6.2
WAP using function that add all prime numbers from 1 to n, n given by the user.

```python
def is_prime(num):
    if num < 2:
        return False
    for j in range(2, num):
        if num % j == 0:
            return False
    return True


def sum_prime(n):
    total = 0
    for i in range(2, n + 1):
        if is_prime(i):
            total = total + i
    return total


n = int(input("Enter n: "))
print("Sum of prime numbers =", sum_prime(n))
```
