## Q.6 Create the employees dataset.

```python
emp_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

emp = pd.DataFrame(emp_data)
emp.to_csv("employees.csv", index=False)

print(emp)
```

**(a)** Shape of the DataFrame.

```python
print("Shape:", emp.shape)
print("Rows:", emp.shape[0], " Columns:", emp.shape[1])
```

**(b)** Summary with data types and non-null counts.

```python
emp.info()
```

**(c)** Descriptive statistics.

```python
print(emp.describe())
```

**(d)** First 5 rows and last 3 rows.

```python
print("First 5 rows:")
print(emp.head())

print("\nLast 3 rows:")
print(emp.tail(3))
```

**(e)** Basic statistics from the dataset.

```python
print("Average salary        :", emp["Salary"].mean())
print("Total bonus paid      :", emp["Bonus"].sum())
print("Youngest employee age :", emp["Age"].min())
print("Highest rating        :", emp["Rating"].max())
```

**(f)** Sort by Salary in descending order.

```python
print(emp.sort_values(by="Salary", ascending=False))
```

**(g)** Categorize employees based on performance rating.

```python
def category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"


emp["Performance"] = emp["Rating"].apply(category)
print(emp[["Name", "Rating", "Performance"]])
```

**(h)** Identify missing values.

```python
print(emp.isnull().sum())
print("\nTotal missing values:", emp.isnull().sum().sum())
```

**(i)** Rename Employee_ID to ID.

```python
emp = emp.rename(columns={"Employee_ID": "ID"})

print(emp.columns.tolist())
```

**(j)** Find employees with more than 5 years of experience, and employees in IT.

```python
print("More than 5 years of experience:")
print(emp[emp["Years_of_Experience"] > 5])

print("\nEmployees in the IT department:")
print(emp[emp["Department"] == "IT"])
```

**(k)** Add a Tax column that deducts 10% of the Salary.

```python
emp["Tax"] = emp["Salary"] * 0.10

print(emp[["Name", "Salary", "Tax"]])
```

**(l)** Save the modified DataFrame to a new CSV file.

```python
emp.to_csv("employees_modified.csv", index=False)

print("File saved successfully.")
print(pd.read_csv("employees_modified.csv"))
```
