### Assignment 6.1WAP using function that add all odd numbers from 1 to n, n is given by the user.

```python
def sum_odd(n):    total = 0    for i in range(1, n + 1):        if i % 2 != 0:            total = total + i    return totaln = int(input("Enter n: "))print("Sum of odd numbers =", sum_odd(n))
```
