## Q.2 Basic NumPy array questions
**a) Reverse the array**

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original:", arr)
print("Reversed:", arr[::-1])
```

**b) Most frequent value and its indices**

Using `np.unique` with `return_counts=True` so that if two values have the same highest count (like in `y`), both get printed.

```python
def most_frequent(a):
    values, counts = np.unique(a, return_counts=True)
    modes = values[counts == counts.max()]
    print("Array:", a)
    for m in modes:
        print(f"  value {m} occurs {counts.max()} times at indices {np.where(a == m)[0]}")

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

print("i.")
most_frequent(x)
print("\nii.")
most_frequent(y)
```

## Q.3 Accessing elements of a 2D array
Indexing starts from 0, so 1st row = index 0, 2nd column = index 1 and so on.

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])
print(arr)

print("a) 1st row, 2nd column:", arr[0, 1])
print("b) 3rd row, 1st column:", arr[2, 0])
```

## Q.4 linspace array named after me

```python
jasroop = np.linspace(10, 100, 25)
print(jasroop)

print("\nDimensions    :", jasroop.ndim)
print("Shape         :", jasroop.shape)
print("Total elements:", jasroop.size)
print("Data type     :", jasroop.dtype)
print("Total bytes   :", jasroop.nbytes)
```

```python
# transpose using reshape -> row of 25 becomes a column (25 x 1)
jasroop_T = jasroop.reshape(25, 1)
print("Shape after reshape:", jasroop_T.shape)
print(jasroop_T)
```

```python
# trying the same with .T
print("Shape with .T:", jasroop.T.shape)
```

**Can we do the same with `.T`?**

No, not directly. A 1D array has only one axis, so `.T` has nothing to swap and returns the same array with shape `(25,)`.
`.T` only works once the array is 2D, e.g. `jasroop.reshape(1, 25).T` gives shape `(25, 1)`.

```python
print(jasroop.reshape(1, 25).T.shape)
```

## Q.5 2D array ucs420_jasroop

```python
ucs420_jasroop = np.array([[10, 20, 30, 40],
                           [50, 60, 70, 80],
                           [90, 15, 20, 35]])
print(ucs420_jasroop)

print("\nMean   :", np.mean(ucs420_jasroop))
print("Median :", np.median(ucs420_jasroop))
print("Max    :", np.max(ucs420_jasroop))
print("Min    :", np.min(ucs420_jasroop))
print("Unique :", np.unique(ucs420_jasroop))
```
